```python
employee_df = pd.DataFrame({
    'Employee': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Department': ['IT', 'HR', 'Sales', 'IT', 'Finance'],
    'Experience': [2, 4, 3, 5, 1]
})

new_df = employee_df[['Employee', 'Experience']]
print(new_df)
```

```python
print(employee_df['Department'])
print(employee_df.columns)
```

```python
employee_df['Office'] = 'Head Office'
print(employee_df)
```

```python
employee_df['Office'] = np.arange(5)
print(employee_df)
```

```python
office_names = pd.Series(
    ['Delhi', 'Mumbai', 'Pune'],
    index=[0, 2, 4]
)

employee_df['Office'] = office_names
print(employee_df)
```

```python
employee_df.drop(columns=['Office'], inplace=True)
print(employee_df)
```

```python
product_df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Electronics'],
    'Price': [50000, 20000, 15000]
})

new_product_df = product_df[['Price', 'Product', 'Category']]
print(new_product_df)
```

```python
df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Rahul'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Mumbai', 'Pune']
})

new_df = df.reindex(columns=['Name', 'Age', 'City', 'Marks'])
print(new_df)
```

```python
marks_df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Rahul'],
    'Total Marks': [85, 92, 78]
})

print(marks_df['Total Marks'])
```

```python
course_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Course': ['DS', 'Java', 'Python', 'DBMS', 'DSA'],
    'Score': [85, 90, 78, 88, 92]
})

print(course_df.columns)

print(course_df['Student'])

new_df = course_df[['Student', 'Score']]
print(new_df)

course_df['Batch'] = '2026'
print(course_df)
```

```python
travel_df = pd.DataFrame({
    'Place': ['Delhi', 'Mumbai', 'Pune', 'Agra', 'Jaipur']
})

travel_df['Code'] = np.arange(5)
print(travel_df)

code_series = pd.Series(['A1', 'A4'], index=[1, 4])

travel_df['Code'] = code_series
print(travel_df)
```

```python
school_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Class': [10, 10, 12, 11, 12],
    'Section': ['A', 'B', 'A', 'C', 'B']
})

new_df = school_df.reindex(
    columns=['Student', 'Class', 'Section', 'House']
)

print(new_df)

new_df['House'] = 'Blue'
print(new_df)
```

```python
inventory_df = pd.DataFrame({
    'Item': ['Pen', 'Book', 'Bag', 'Pencil', 'Eraser'],
    'Quantity': [10, 20, 5, 15, 8],
    'Price': [10, 50, 500, 5, 3]
})

print(inventory_df)

stock = pd.Series(
    [100, 200, 300],
    index=[0, 2, 4]
)

inventory_df['Stock'] = stock
print(inventory_df)

inventory_df.drop(columns=['Stock'], inplace=True)
print(inventory_df)
```
